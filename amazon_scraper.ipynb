## **Web Scraping Amazon Data**
 
The Objective of the Assignment is to develop a Python-based web scraping solution that fetch laptop product information from Amazon India (amazon.in) 
such as Product Image URL,
Product Title,
Product Rating,
Product Price,
Product Type (Advertisement or Organic Result) 
and store the extracted data in a structured CSV file.


```python
# Required Libraries
from bs4 import BeautifulSoup
import pandas as pd
import requests
import time
from datetime import datetime

print("All libraries imported successfully!")
```

#### **Retrieve and Parse HTML Content from Amazon**

```python
headers = {
    "User-Agent": (
        "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
        "AppleWebKit/537.36 (KHTML, like Gecko) "
        "Chrome/137.0.0.0 Safari/537.36"
    ),
    "Accept-Language": "en-IN,en;q=0.9"
}

all_products = []

for page_num in range(1, 6):

    url = f"https://www.amazon.in/s?k=laptop&page={page_num}"

    response = requests.get(url, headers=headers)

    soup = BeautifulSoup(response.text, "html.parser")

    products = soup.select('div[data-component-type="s-search-result"]')

    all_products.extend(products)

print("Total Products:", len(all_products))
```

#### **Extract Product Information**

**Laptop Title**

```python
product_title = []

for product in all_products:

    title_tag = product.select_one("h2")

    title = title_tag.get_text(strip=True) if title_tag else "N/A"

    product_title.append(title)

print(product_title)
```

Laptop's Rating

```python
all_prices = []
for product in all_products:

    price_tag = product.select_one("span.a-price-whole")

    if price_tag:
        price = price_tag.get_text(strip=True)
    else:
        price = "NA"

    all_prices.append(price)
    print(price)

time.sleep(2)
```

Rating

```python
Product_ratings = []
for product in all_products:

    rating_tag = product.select_one("span.a-size-small.a-color-base")

    if rating_tag:
        rating = rating_tag.get_text(strip=True)
    else:
        rating = "NA"

    Product_ratings.append(rating)
    print(rating)
```

Laptop's Image RUL

```python
images_urls = []
for product in all_products:

    image_tag = product.select_one("img.s-image")

    if image_tag and image_tag.has_attr("src"):
        image_url = image_tag["src"]
    else:
        image_url = "NA"

    images_urls.append(image_url)
    print(image_url)
```

Product Type (Ad/Organic)

```python
product_type = []

for product in all_products:

    type_tag = product.select_one("span.a-color-secondary")

    p_type = "Ad" if type_tag and "Sponsored" in type_tag.get_text() else "Organic"

    product_type.append(p_type)

print(product_type)
```

**Create DataFrame**

```python
df = pd.DataFrame({
    "Title": product_title,
    "Price": all_prices,
    "Rating": Product_ratings,
    "Image": images_urls,
    "Product_Type(Ad/Organic)": product_type
})

# Display first 5 rows
df.head()
```

```python
print("Total Records:", len(df))
```

**Save the CSV with a timestamp**

```python
timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")

# output file name
Amazon_laptop_data = f"amazon_laptop_data_{timestamp}.csv"

# Save DataFrame
df.to_csv(Amazon_laptop_data, index=False)

print(Amazon_laptop_data)
```
